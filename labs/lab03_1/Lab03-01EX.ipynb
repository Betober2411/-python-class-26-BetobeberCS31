### Format Specifier


``.2f`` is a **format specifier** that controls how a number is displayed inside an f-string. It means: show this number as a **float with exactly 2 digits after the decimal point**.

The pattern is {variable:format}, and .2f breaks down like this:

- .2 means 2 digits after the decimal point.
- f means fixed-point (float) notation.

What it does to your three variables


|Expression|	Value stored|	Displayed as|
|----|-----|-|
|{miles_driven:.2f}|	150.0	|150.00|
|{gallons_used:.2f}|	15.2	|15.20|
|{cost_per_gallon:.2f}|	4.25|	4.25|

#### Rounding

``.2f`` also rounds the value if it has more decimals:

x = 3.14159
print(f"{x:.2f}")   #3.14
print(f"{x:.3f}")   #3.142
print(f"{x:.0f}")   #3

## Indentation

In Python, indentation defines which lines belong together in a block. Other languages use braces ``{ }`` for this. Python uses the colon ``:`` to open a block and the indentation to show what is inside it. A line that is indented more belongs to the line above it that ends in ``:``.

### Rules to remember
- A line ending in : (``while``, ``if``, ``elif``, ``else``, ``for``, ``def``) must be followed by an indented block.
- The standard is 4 spaces per level. IDLE and VS Code do this when you press Tab.
- Every line in the same block must have exactly the same indentation. Inconsistent spacing causes IndentationError.
- To end a block, simply dedent back to the level of the line that opened it (or an outer one).

In the structure. else: lines up with if and elif, because they are one connected statement. Only the **body** of each branch is indented. So the messages under if/elif and the calculations under else are all at the **same level (Level 2)**.

#### Why not dedent all the way to Level 0?

If ``choice = input("Get entries for another trip...")`` were at Level 0, it would sit **outside** the loop. Python would run the loop forever (since ``choice`` never changes inside it) or ask the question only once after the loop ends. The last line, ``print("Thank you!!")``, is at Level 0 on purpose: it runs **once**, after the loop finishes.
