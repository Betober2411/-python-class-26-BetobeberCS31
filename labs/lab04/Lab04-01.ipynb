```python
# Guess the number Program 
# Ask for player's name. Output their name at least once.
# Ask player if they would like to play an EASY, MODERATE, or HARD game. 
#       - easy = range of 1 to 10, limit tries to 5.
#       - medium = range of 1 to 100, limit tries to 8
#       - hard = range of 1 to 1000, limit tries to 10    
# If the player doesn't guess the number in the required amount of tries, they lose the game
# Keep track of the wins and loses for the player, displaying the score before asking if they would like to play again.
# be sure to include comments.

import random
```

## 1. Displays the game title

```python
# ---------- Displays the game title ----------
def display_title():
    print("Guess the number!")
    print()
```

## 2. Asks for the player's name, greets them, and RETURNS it 

```python
# ---------- Asks for the player's name, greets them, and RETURNS it ----------
def get_name():
    player_name = input("What is your name?  ")
    print("Hello", player_name)
    print()
    return player_name          # CHANGED: return the name so main() can reuse it
```

## 3. Asks for a difficulty level until the input is valid

```python
# ---------- Asks for a difficulty level until the input is valid ----------
# Returns a tuple: (highest number in range, number of attempts allowed)
def get_level():                # NEW: split out of play_game()
    while True:                 # repeat until we get a valid choice
        level = input("What game level would you like to play? "
                      "E for Easy, M for Medium, or H for Hard: ").lower()
        if level == "e":
            return 10, 5
        elif level == "m":
            return 100, 8
        elif level == "h":
            return 1000, 10
        else:
            print("Sorry. Invalid input. Please try again.")
```

## 5. Plays ONE round. Returns True if the player won, False if lost

```python
# ---------- Plays ONE round. Returns True if the player won, False if lost ----------
def play_game():                # CHANGED: no longer takes or returns 'wins'
    max_number, tries = get_level()          # CHANGED: level logic moved out
    number = random.randint(1, max_number)
    print("I'm thinking of a number from 1 to " + str(max_number) + "\n")

    count = 1
    while count <= tries:
        try:                                  # NEW: guard against non-numeric input
            guess = int(input("Your guess: "))
        except ValueError:
            print("Please enter a whole number.")
            continue                          # doesn't use up an attempt

        if guess < number:
            print("Too low.")
        elif guess > number:
            print("Too high.")
        else:
            print("You guessed it in " + str(count) + " tries.\n")
            return True                       # CHANGED: report a win
        count += 1                            # CHANGED: one increment instead of three

    # Reaching here means the loop ended with no correct guess
    print("Sorry. You ran out of guesses. The number was", number)
    return False
```

## 6. Runs the whole program and keeps the score

```python
# ---------- Runs the whole program and keeps the score ----------
def main():
    display_title()
    player_name = get_name()                  # CHANGED: keep the returned name
    wins = 0                                  # initialize score counters
    losses = 0                                # NEW
    again = "y"
    while again.lower() == "y":
        won = play_game()                     # CHANGED: receives True/False
        if won:
            wins += 1
        else:
            losses += 1
        # NEW: scoreboard after every round, before asking to play again
        print("--- Scoreboard for " + player_name + " ---")
        print("Wins:", wins, "| Losses:", losses)
        print()
        again = input("Would you like to play again? (y/n): ")
        print()
    print("Thank you for playing, " + player_name + "!!")   # CHANGED: name shown again
```

```python
#if started as the main module, call the main function
if __name__ == "__main__":
    main() 
```
